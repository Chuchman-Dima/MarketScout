```python
import csv
import pandas as pd

df = pd.read_csv("new_cars_dataset_2.csv",low_memory=False)

# Примусово беремо всі нечислові поля в подвійні лапки
df.to_csv(
    "cars_strict.csv",
    sep=",",
    index=False,
    encoding="utf-8-sig",
    quoting=csv.QUOTE_NONNUMERIC,  # жодна кома всередині тексту не розіб'є рядок
)
```

```python
import re

# 1. Словник стандартизованих англійських назв для міст України
CITY_TRANSLATION_MAP = {
    "Київ": "Kyiv",
    "Вінниця": "Vinnytsia",
    "Дніпро": "Dnipro",
    "Житомир": "Zhytomyr",
    "Запоріжжя": "Zaporizhzhia",
    "Івано-Франківськ": "Ivano-Frankivsk",
    "Кропивницький": "Kropyvnytskyi",
    "Луцьк": "Lutsk",
    "Львів": "Lviv",
    "Миколаїв": "Mykolaiv",
    "Одеса": "Odesa",
    "Полтава": "Poltava",
    "Рівне": "Rivne",
    "Суми": "Sumy",
    "Тернопіль": "Ternopil",
    "Ужгород": "Uzhhorod",
    "Харків": "Kharkiv",
    "Херсон": "Kherson",
    "Хмельницький": "Khmelnytskyi",
    "Черкаси": "Cherkasy",
    "Чернівці": "Chernivtsi",
    "Чернігів": "Chernihiv",
    "Лисичанськ": "Lysychansk",
    "Кадіївка": "Kadiivka",
    "Марківка": "Markivka",
    "Кривий Ріг": "Kryvyi Rih",
    "Маріуполь": "Mariupol",
    "Кременчук": "Kremenchuk",
    "Біла Церква": "Bila Tserkva",
    "Мелітополь": "Melitopol",
    "Нікополь": "Nikopol",
    "Бердянськ": "Berdiansk",
    "Кам'янець-Подільський": "Kamianets-Podilskyi",
    "Бровари": "Brovary",
    "Перевальськ": "Perevalsk",
    "Сєвєродонецьк": "Sievierodonetsk",
    "Рубіжне": "Rubizhne",
    "Старобільськ": "Starobilsk",
    "Первомайськ": "Pervomaisk",
    "Хрустальний": "Khrustalnyi",
    "Ровеньки": "Rovenky",
}


def clean_and_translate_city(val):
    if pd.isna(val):
        return None

    val = str(val).strip()

    # Відкидаємо записи областей (наприклад, "Донецька обл.")
    if "обл." in val.lower() or "область" in val.lower():
        return None

    # Прибираємо текст у дужках: "Дніпро (Дніпропетровськ)" -> "Дніпро"
    cleaned = re.sub(r"\(.*?\)", "", val).strip()

    # Повертаємо англійський відповідник зі словника
    return CITY_TRANSLATION_MAP.get(cleaned, cleaned)


# Припускаємо, що колонка з містом називається 'City' (якщо в датасеті інша назва, наприклад 'city_name', замініть тут)
city_col = "City" if "City" in df.columns else "city_name"

# Створюємо чисту англомовну колонку
df["City_EN"] = df[city_col].apply(clean_and_translate_city)

print("Унікальні перекладені міста:")
print(df["City_EN"].dropna().unique())
```

```python
import csv

df.to_csv(
    "cars_strict.csv",
    sep=",",
    index=False,
    encoding="utf-8-sig",
    quoting=csv.QUOTE_NONNUMERIC,
)
```
